```python
import os
from openai import OpenAI
```

```python
client = OpenAI(
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url="https://api.deepseek.com",
)
```

```python
response = client.chat.completions.create(
  model="deepseek-chat",
  messages=[
    {"role": "user", "content": "四大文明古国分别有哪些"}
  ]
)
response
```

```python
print(response.choices[0].message.content)
```
