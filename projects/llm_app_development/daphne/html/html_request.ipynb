```python
import requests
head = {'User-Agent': 'Mozilla/5.0'}
res = requests.get('http://books.toscrape.com', headers=head)
if res.ok:
    print(res.text)
else:
    print('Failed to retrieve the page')
```

```python
import requests
headers = {'User-Agent': "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/148.0.0.0 Safari/537.36 Edg/148.0.0.0"}
response = requests.get("https://movie.douban.com/top250", headers=headers)
print(response.status_code)  # https://developer.mozilla.org/zh-CN/docs/Web/HTTP/Reference/Status/418
print(response.text)
```
