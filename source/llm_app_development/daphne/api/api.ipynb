#### class\05 API用法示例\02 AI大模型API用法示例 _ 文本撰写：秒生成小红书爆款文案.ipynb

system_prompt 是给模型洗脑的宪法，user_prompt 是让模型处理的具体案件。assistant 是模型给出的判决书与历史档案。这三者必须分开存放，才能确保模型既“听话”，又“办对事”，同时还能记得“之前判过什么案”，实现连贯的多轮对话。

````python
def get_openai_response(client, system_prompt, user_prompt, assistant_prompt, model="qwen-turbo"):
    response =  client.chat.completions.create(
        model=model,
        messages=[{"role":"system","content":system_prompt},
                  {"role":"user","content":user_prompt},
                  {"role":"assistant","content":assistant_prompt}
                  ]
    )
    return response.choices[0].message.content
````
