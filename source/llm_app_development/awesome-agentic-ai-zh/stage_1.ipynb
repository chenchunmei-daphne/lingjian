##### 基于规则的方法

```python
def classify_intent_rule(query: str) -> str:
    query = query.lower()
    intent_rules = {
        "报销流程查询": ['报销', "费用", "报销流程"],
        "保险产品销售技巧": ['卖保险', '销售', '销售技巧', '技巧'],
    }  
    for intent, keywords in intent_rules.items():
        if any(keyword in query for keyword in keywords):
            return intent
    return "未知意图"
```

```python
query = ["怎么报销保险费用？", "保险销售技巧有哪些？", "保险理赔需要哪些资料？"]
for q in query:
    intent = classify_intent_rule(q)
    print(f"Query: {q} -> Intent: {intent}")
```

##### 基于机器学习的方法（BERT 分类模型）

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
```

```python
# from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

# 使用一个现成的中文情感分类模型（5分类，评价星级）
# 加载一个分词器（Tokenizer），该分词器与指定的预训练模型配套。
tokenizer = AutoTokenizer.from_pretrained("nlptown/bert-base-multilingual-uncased-sentiment")

# 加载一个序列分类模型（Sequence Classification），即用于文本分类任务的预训练模型。
model = AutoModelForSequenceClassification.from_pretrained("nlptown/bert-base-multilingual-uncased-sentiment")

# 测试
query = "怎么报销保险费费用？"
inputs = tokenizer(query, return_tensors="pt") # 将原始文本通过分词器处理成模型可接受的输入格式,返回字典，return_tensors="pt" 表示返回的数据类型是 PyTorch 张量（Tensor）
outputs = model(**inputs)
pred_id = outputs.logits.argmax(dim=1).item()  # 从模型的原始输出中取出分数最高的类别ID，作为最终预测结果。

# 这个模型是1-5星评价，1星最负面，5星最正面
print(f"查询: {query} -> 情感评分: {pred_id + 1} 星")
```
